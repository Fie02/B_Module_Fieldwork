```python
import math
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df_lake=pd.read_excel("Measurements_Fie.xlsx", sheet_name="LG_lake")
df_spring=pd.read_excel("Measurements_Fie.xlsx", sheet_name="LG_spring")
df_S3=pd.read_excel("Measurements_Fie.xlsx", sheet_name="S3")
df_S4=pd.read_excel("Measurements_Fie.xlsx", sheet_name="S4")
```

```python
df_spring
```

```python
def plot_pH(df, loca='LG_lake'):

    fines=[0,1,2]
    meds=[3,4,5]
    coarse=[6,7,8]

    CB_color_cycle = ['#377eb8', '#ff7f00', '#4daf4a',
                '#f781bf', '#a65628', '#984ea3',
                '#999999', '#e41a1c','#dede00']

    plt.scatter(df.loc[fines, 'ph_0'], df.loc[fines, 'ph_10'], marker='x', color=CB_color_cycle[0], label='Fine grains')
    plt.scatter(df.loc[meds, 'ph_0'], df.loc[meds, 'ph_10'], marker='o', color=CB_color_cycle[1], label='medium grains')
    plt.scatter(df.loc[coarse, 'ph_0'], df.loc[coarse, 'ph_10'], marker='d', color=CB_color_cycle[2], label='Coarse grains')

    plt.xlabel('pH at t=0 min.')
    plt.ylabel('pH at t=10 min.')
    plt.grid()
    plt.legend(loc='right', bbox_to_anchor=(1.4,0.1))
    plt.savefig(f'updates_fie/{loca}_pH_start_end.png', bbox_inches='tight', dpi=300)
```

```python
plot_pH(df_lake, loca='LG_lake')
```

```python
plot_pH(df_spring, loca='LG_spring')
```

```python
plot_pH(df_S3, loca='S3')
```

```python
plot_pH(df_S4, loca='S4')
```
